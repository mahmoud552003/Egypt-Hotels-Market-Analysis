```python
import numpy as np
import re

def convert_distance(value):
    if pd.isna(value):
        return np.nan

    value = value.lower()

    number = float(re.search(r'(\d+\.?\d*)', value).group())

    if "km" in value:
        return number
    elif "m" in value:
        return number / 1000

    return np.nan

df['distance_from_center'] = df['distance_from_center'].apply(convert_distance)
```
